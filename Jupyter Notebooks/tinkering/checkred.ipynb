```python
import numpy as np
import sqlite3
import pandas
import requests
import json
import csv
```

```python
r = requests.get("https://icecube.wisc.edu/~grbweb_public/GRBweb2.sqlite")
f = open('GRBweb2.sqlite', 'wb').write(r.content)
db = sqlite3.connect('GRBweb2.sqlite')
table_names = pandas.read_sql_query("SELECT * from sqlite_sequence", db)
```

```python
red_table = pandas.read_sql_query("SELECT * from Summary", db)
red_table = red_table.sort_values("redshift")
final_table = red_table[:687]
#red_table.iloc[8].to_numpy()[1] #name
#red_table.iloc[8].to_numpy()[2] #id
#red_table.iloc[8].to_numpy()[17] #redshift
#red_table.iloc[8].to_numpy()[18] #redshift source
#red_table.iloc[8].to_numpy()[19] #T100
output_fermi = {}
output_other = {}
fermi_id = []
for i in range(0, len(final_table)):
    array = final_table.iloc[i].to_numpy()
    grb_id = array[2]
    if grb_id == None:
        next_other = {"Name": array[1], "id": array[4], "redshift": array[17], "redshift_source": array[18], "T100": array[19]}
        output_other[array[1]] = next_other
    else:
        id_stripped = str(grb_id).split("B")[1]
        next_fermi = {"Name": array[1], "id": id_stripped, "redshift": array[17], "redshift_source": array[18], "T100": array[19]} 
        output_fermi[id_stripped] = next_fermi
        fermi_id.append({"id": id_stripped})
    #red_index = red_table.iloc[i].to_numpy()[0] - 1
    #red_val = red_table.iloc[i].redshift.get(red_index)
    #red_source = str(red_table.iloc[i].redshift_source.get(red_index))
    #t100 = red_table.iloc[i].T100.get(red_index)

other_str = json.dumps(output_other, indent=4)
with open("non_fermi.json", "w") as f:
    f.write(other_str)

fermi_str = json.dumps(output_fermi, indent=4)
with open("fermi.json", "w") as f:
    f.write(fermi_str)

print(other_str)
print(fermi_str)

with open('fermi_ids.csv', 'w', newline='') as csvfile:
    fieldnames = ['id']
    writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
    writer.writeheader()
    writer.writerows(fermi_id)
```

```python
def checkRed(fermiID):
    name_Fermi = "GRB" + fermiID
    t100 = 0
    red_val = 0
    red_source = ""
    try:
        check_table = pandas.read_sql_query("SELECT * from Summary", db)
        check_table = check_table.sort_values("GRB_name")
        indices = np.where(check_table == name_Fermi)
        row_indices, col_indices = indices[0], indices[1]
        red_index = check_table.iloc[row_indices].to_numpy()[0][0] - 1
        red_val = check_table.iloc[row_indices].redshift.get(red_index)
        red_source = str(check_table.iloc[row_indices].redshift_source.get(red_index))
        t100 = check_table.iloc[row_indices].T100.get(red_index)
    finally:
        return [name_Fermi, t100, red_val, red_source]
```

```python
def checkID(ID, return_object):
    result = []
    for i in range(0, 999):
        addendum = str(i)
        if i < 100:
            addendum = "0" + str(i)
        elif i < 10:
            addendum = "00" + str(i)
        name_Fermi = str(ID) + addendum

        output = checkRed(name_Fermi)
        if output[1] != 0:
            result = output
    
    if result[1] != 0:
        return_object[result[0]] = [result[1],result[2],result[3]]
        return return_object
    else:
        return "Error 1: checkFailed"
```
