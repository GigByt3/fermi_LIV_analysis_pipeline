```python
from gdt.missions.fermi.gbm.collection import GbmDetectorCollection
from gdt.missions.fermi.gbm.phaii import Cspec
from gdt.core.plot.lightcurve import Lightcurve
import matplotlib.pyplot as plt
```

```python
cspec_n0 = Cspec.open("data/glg_cspec_n0_bn160118060_v00.pha")
cspec_n1 = Cspec.open("data/glg_cspec_n1_bn160118060_v00.pha")
cspec_n2 = Cspec.open("data/glg_cspec_n2_bn160118060_v00.pha")
```

```python
cspecs = GbmDetectorCollection.from_list([cspec_n0, cspec_n1, cspec_n2])
cspecs
```

```python
cspec_n0
```

```python
cspecs.energy_range()
```

```python
cspecs.to_lightcurve(time_range=[-10.0, 10.0])
```

```python
lcplot = Lightcurve(cspecs.to_lightcurve(time_range=[-10.0, 10.0])[1])
plt.xlim(-10,30)
plt.show()
```

```python
nai_kwargs = {'energy_range': (8.0, 100.0)}
lcplot = Lightcurve(cspecs.to_lightcurve(time_range=[-10.0, 10.0], nai_kwargs=nai_kwargs)[1])
plt.xlim(-10,30)
plt.show()
```
